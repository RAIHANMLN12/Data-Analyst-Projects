## 1. Data Understanding

```python
import pandas as pd

dataset = pd.read_csv('data/marketing_AB.csv')

dataset.head(10)
```

```python
dataset.info()
```

## 2. Calculate conversion rate per group

```python
conversion_rate = pd.crosstab(
    dataset["test group"],
    dataset["converted"],
    normalize="index"
).mul(100)

conversion_rate.map(lambda x: f"{x:.2f}%")
```

## 3. Organize the data into a contingency table

```python
contingency_table = pd.crosstab(
    dataset["test group"],
    dataset["converted"]
)

contingency_table
```

## 4. Run the significance test

```python
from scipy.stats import chi2_contingency

# Perform the Chi-Square test
chi2_stat, p_value, dof, expected = chi2_contingency(contingency_table)

print("Chi-square statistic:", round(chi2_stat, 4))
print("P-value:", round(p_value, 4))

# Significance level
alpha = 0.05

if p_value < alpha:
    print("Reject H0. There is a significant difference in conversion between ad and psa.")
else:
    print("Fail to reject H0. There is not enough evidence of a difference in conversion between ad and psa.")
```

## 5. Calculate the confidence interval

```python
import numpy as np
from statsmodels.stats.proportion import confint_proportions_2indep

# Ambil jumlah converted dan total dari masing-masing group
ad_converted = contingency_table.loc["ad", True]
ad_total = contingency_table.loc["ad"].sum()

psa_converted = contingency_table.loc["psa", True]
psa_total = contingency_table.loc["psa"].sum()


# Hitung conversion rate
ad_rate = ad_converted / ad_total
psa_rate = psa_converted / psa_total

# Selisih conversion rate
difference = ad_rate - psa_rate


# Hitung 95% confidence interval
ci_low, ci_high = confint_proportions_2indep(
    ad_converted,
    ad_total,
    psa_converted,
    psa_total,
    compare="diff",
    alpha=0.05,
    method="wald"
)


# Tampilkan hasil
print("Ad conversion rate :", f"{ad_rate * 100:.2f}%")
print("PSA conversion rate:", f"{psa_rate * 100:.2f}%")

print(
    "Difference in conversion rate:",
    f"{difference * 100:.2f} percentage points"
)

print(
    "95% Confidence Interval:",
    f"[{ci_low * 100:.2f}%, {ci_high * 100:.2f}%]"
)
```

## 6. Check efficiency, not just raw effectiveness

```python
efficiency = dataset.groupby("test group").agg(
    total_users=("converted", "count"),
    conversions=("converted", "sum"),
    total_ads=("total ads", "sum")
)

# Conversion rate
efficiency["conversion_rate"] = (
    efficiency["conversions"] / efficiency["total_users"]
)

# Rata-rata ads yang dilihat setiap orang
efficiency["avg_ads_per_person"] = (
    efficiency["total_ads"] / efficiency["total_users"]
)

# Jumlah conversion per 1,000 ads
efficiency["conversions_per_1000_ads"] = (
    efficiency["conversions"]
    / efficiency["total_ads"]
    * 1000
)

efficiency
```

## 7. Visualization

```python
import matplotlib.pyplot as plt
from statsmodels.stats.proportion import proportion_confint

# Confidence interval untuk Ad
ad_ci_low, ad_ci_high = proportion_confint(
    count=ad_converted,
    nobs=ad_total,
    alpha=0.05,
    method="wilson"
)

# Confidence interval untuk PSA
psa_ci_low, psa_ci_high = proportion_confint(
    count=psa_converted,
    nobs=psa_total,
    alpha=0.05,
    method="wilson"
)


# Conversion rate
conversion_rates = [
    ad_rate * 100,
    psa_rate * 100
]

# Error bar
lower_error = [
    (ad_rate - ad_ci_low) * 100,
    (psa_rate - psa_ci_low) * 100
]

upper_error = [
    (ad_ci_high - ad_rate) * 100,
    (psa_ci_high - psa_rate) * 100
]


# Visualisasi
groups = ["Ad", "PSA"]

plt.figure(figsize=(8, 5))

plt.bar(
    groups,
    conversion_rates,
    yerr=[lower_error, upper_error],
    capsize=6
)

plt.ylabel("Conversion Rate (%)")
plt.xlabel("Test Group")
plt.title("Conversion Rate by Test Group with 95% Confidence Interval")

plt.show()
```
